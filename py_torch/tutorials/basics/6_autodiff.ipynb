```python
x = torch.ones(5)
y = torch.zeros(3)
w = torch.randn(5, 3, requires_grad=True)
b = torch.randn(3, requires_grad=True)
z = x @ w + b
loss = torch.nn.functional.binary_cross_entropy_with_logits(z, y)
```

```python
z.grad_fn
```

```python
loss.grad_fn
```

```python
loss.backward()
w.grad
```

```python
b.grad
```

```python
z = torch.matmul(x, w) + b
print(z.requires_grad)

with torch.no_grad():
    z = torch.matmul(x, w) + b
print(z.requires_grad)
```

```python
# Same as 
z = torch.matmul(x, w) + b
z_det = z.detach()
z_det.requires_grad
```

```python
inp = torch.eye(4, 5, requires_grad=True)
out = (inp + 1).pow(2).t()
out.backward(torch.ones_like(out), retain_graph=True)
print(inp.grad)
```

```python
out.backward(torch.ones_like(out), retain_graph=True)
inp.grad
```

```python
inp.grad.zero_()
out.backward(torch.ones_like(out), retain_graph=True)
inp.grad
```
